```python
fileName = 'input.txt'
globalInput = 'you'
globalOutput = 'out'

def read_input (fileName) :
    print(f'Reading file : {fileName}')
    file = open(fileName, 'r').read().split("\n")
    rack = {}
    for line in file :
        input_output = line.split(':')
        rack[input_output[0]] = input_output[1].split()
    return rack

def find_paths_to_reach_out() :
    # Without making loop max len of path is max len of rack
   maxExplorationCount = len(serverRack)
   explore_input(globalInput, maxExplorationCount)

def explore_input(input, explorationDepth) :
    global pathToGlobalOutputFound
    nextExplorationDepth = explorationDepth - 1
    if (nextExplorationDepth < 0) :
        print(f'Maximum depth is reached')
        return
    for output in serverRack[input] :
        if output == globalOutput :
            pathToGlobalOutputFound += 1
            print(f'Output find')
        else :
            print(f'Explore {output} from {input}:{serverRack[input]}')
            explore_input(output, nextExplorationDepth)

serverRack = read_input(fileName)
pathToGlobalOutputFound = 0
print('server rack', serverRack)
find_paths_to_reach_out()

print(f'Found {pathToGlobalOutputFound} paths from \'{globalInput}\' to \'{globalOutput}\'')
```

```python
fileName = 'input.txt'
globalInput = 'svr'
globalOutput = 'out'

def read_input (fileName) :
    print(f'\n Reading file : {fileName}')
    file = open(fileName, 'r').read().split("\n")
    rack = {}
    components = {}
    inversedRack = {}
    for line in file :
        input_output = line.split(':')
        inputValue = input_output[0]
        outputValues = input_output[1].split()
        rack[inputValue] = {}
        for output in outputValues :
            rack[inputValue].update({output: ''})
            if inversedRack.get(output) == None :
                inversedRack[output] = [inputValue]
            else :
                inversedRack[output].append(inputValue)
        components[inputValue] = 0
    print(f'End reading file\n')
    return rack, components, inversedRack

def find_paths_to_reach_out() :
    # Without making loop max len of path is max len of rack
   maxExplorationCount = len(serverRack)
   print(f'Max exploration depth {maxExplorationCount}')
   explore_input(globalInput, maxExplorationCount)

def explore_input(input, explorationDepth) :
    global problematicPathFound
    global rackComponents
    if (explorationDepth - 1 < 0) :
        print(f'Maximum depth is reached')
        return 0
    if (rackComponents[input] != 0) :
        print(f'Find a loop on {input}')
        return 0
    rackComponents[input] = 1
    for output in serverRack[input] :
        if output == globalOutput :
            # print(f'Output find input {input}, output {serverRack[input]}')
            if is_passed_by_problematic_components() :
                print(f'Problematic path found : {rackComponents}')
                problematicPathFound += 1
        else :
            # print(f'Explore {output} from {input}:{serverRack[input]}')
            # if (explore_input(output, explorationDepth - 1) == 0) :
            #     return 0
            explore_input(output, explorationDepth - 1)
    rackComponents[input] = 0
    return

def is_passed_by_problematic_components() :
    for component in problematicComponents :
        if rackComponents[component] != 1 :
            # print(f'Not passed by {component}')
            return False
    # print(f'Passed by {problematicComponents}')
    return True

def compress_server_rack(rack, reversedRack, rackComponents) :
    print('\nStart compressing server rack')
    resultingRack = rack.copy()
    resultingReversedRack = reversedRack.copy()
    for component in rack :
        if len(resultingRack[component]) == 1 :
            # print(f'Server rack snpashot = {resultingRack}')
            # print(f'Server reversed rack snpashot = {resultingReversedRack}')
            for replacingComponent in rack[component] :
                # Use to extract name of the single component
                if not(component == 'svr') and not(component in problematicComponents) and not(replacingComponent == 'out') :
                    resultingRack, resultingReversedRack, rackComponents = replace_component(component, replacingComponent, resultingRack, resultingReversedRack, rackComponents)
    print('End compressing server rack\n')
    return resultingRack, rackComponents


def replace_component (componentToReplace, replacingComponent, normalRack, reversedRack, rackComponents) :
    # print(f'Replacing : {componentToReplace} for each in {reversedRack[componentToReplace]}')
    for input in reversedRack[componentToReplace] :
        # Replace component for rack output
        # print(f'Replace in rack \'{input}\' : {normalRack[input]} the component \'{componentToReplace}\' by {replacingComponent}')
        del normalRack[input][componentToReplace]
        normalRack[input].update({replacingComponent: ''})
        
        # Update reversedRack in consequences
        # print(f'Replace in reversed rack \'{replacingComponent}\' : {reversedRack[replacingComponent]} the component \'{componentToReplace}\' by {input}')
        reversedRack[replacingComponent].append(input)
        # print(f'Results :\n - In rack \'{input}\':{normalRack[input]}\n - In reversed rack \'{replacingComponent}\':{reversedRack[replacingComponent]}\n ---------------')
    reversedRack[replacingComponent].remove(componentToReplace)
    del normalRack[componentToReplace]
    del reversedRack[componentToReplace]
    del rackComponents[componentToReplace]
    return normalRack, reversedRack, rackComponents

    

serverRack, rackComponents, inversedRack = read_input(fileName)
problematicPathFound = 0
problematicComponents = ['dac', 'fft']

componentCount = len(rackComponents)
# print('Server rack', serverRack)
print(f'{componentCount} components : {rackComponents}')
serverRack, rackComponents = compress_server_rack(serverRack, inversedRack, rackComponents)
print('Server rack', serverRack)
print(f'Removing {componentCount-len(rackComponents)} components on {componentCount}')

find_paths_to_reach_out()

print(f'Found {problematicPathFound} paths from \'{globalInput}\' to \'{globalOutput}\'')
```

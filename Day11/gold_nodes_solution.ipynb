```python
fileName = 'gold_small_input.txt'
globalInput = 'svr'
globalOutput = 'out'

class node:
  def __init__(self, data):
    self.value = data
    self.previous = None

def read_input (fileName) :
    print(f'Reading file : {fileName}')
    file = open(fileName, 'r').read().split("\n")
    rack = {}
    for line in file :
        input_output = line.split(':')
        rack[input_output[0]] = input_output[1].split()
    return rack

def find_paths_to_reach_out() :
    # Without making loop max len of path is max len of rack
   maxExplorationCount = len(serverRack)
   
   explore_input(globalInput, maxExplorationCount, None)

def explore_input(input, explorationDepth, previousNode) :
    global problematicPathFound
    nextExplorationDepth = explorationDepth - 1
    headNode = node(input)
    headNode.previous = previousNode
    if (nextExplorationDepth < 0) :
        print(f'Maximum depth is reached')
        return
    for output in serverRack[input] :
        if output == globalOutput :
            # print(f'Output find')
            print(f'Output find ', end='')
            print_previous_chain_nodes(headNode)
            if is_passed_by_components(headNode, ['dac', 'fft']) :
                # print(f'From path to reach {globalOutput} : {fromPath}')
                problematicPathFound += 1
        else :
            # print(f'Explore input {output} from {input}:{serverRack[input]}')
            explore_input(output, nextExplorationDepth, headNode)
    del headNode

def print_previous_chain_nodes(head) :
    pointingNode = head
    print(f'head({pointingNode.value})', end='')
    while (pointingNode.previous != None) :
        pointingNode = pointingNode.previous
        print(f'->{pointingNode.value}', end='')
    print()

def is_passed_by_components(headNode, targetComponents) :
    reachedTargetComponents = [0 for c in targetComponents]

    pointingNode = headNode
    while (pointingNode.previous != None) :
        for i in range(len(targetComponents)) :
            if pointingNode.value == targetComponents[i] :
                reachedTargetComponents[i] += 1
        pointingNode = pointingNode.previous
    # print(f'First node [value: {pointingNode.value}, previous: {pointingNode.previous}]')
        
    for i in range(len(targetComponents)) :
        if reachedTargetComponents[i] == 0 :
            return False
    return True
        

serverRack = read_input(fileName)
problematicPathFound = 0
print('server rack', serverRack)
find_paths_to_reach_out()

print(f'Found {problematicPathFound} paths from \'{globalInput}\' to \'{globalOutput}\'')
```

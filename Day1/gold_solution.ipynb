```python
file_name = 'input.txt'
max_dial = 99
position_possibilities = 100
current_position = 50
intrance_code = 0
instructions = []
print("Start at ", current_position)

def rotate (pointer, direction, tics) :
    global intrance_code
    if (direction == 'L') :
        tranlsation = - int(tics)
        # print("rotate left", tranlsation)
    elif (direction == 'R') :
        tranlsation = int(tics)
        # print("rotate right", tranlsation)
    else :
        print('parse error')
        return False

    pointer_result = (pointer + tranlsation) % position_possibilities

    passing_by_0_times = (pointer + tranlsation) // position_possibilities
    # We need to adjust passing by 0 count when we start or end on 0
    # // operator count if we pass 100 or if you go under 0
    # so if we pass by 100 (R42 from 90) we count 1 time passing by 0
    # if we go under 0 (L60 from 50) we count -1
    # but oposite thinks will not be countted
    if tranlsation < 0 and pointer == 0:
        passing_by_0_times += 1
    if tranlsation > 0 and pointer_result == 0:
        passing_by_0_times -= 1

    intrance_code += abs(passing_by_0_times)
    # print(direction, tranlsation,'time on 0',abs(passing_by_0_times)
    if(current_position == 0) :
        intrance_code += 1 
    return pointer_result


file = open(file_name, 'r').read().split("\n\n")
for line in file:
    instructions = line.split('\n')


for rotation in instructions:
    direction = rotation[0]
    tics = rotation[1:]
    # print("Rotation à " + direction + " de " + tics)
    current_position = rotate (current_position, direction, tics)
    # print('position', current_position)

print("Final position", current_position)
print("Code :", intrance_code)
 
```

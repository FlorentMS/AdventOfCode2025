```python
file_name = 'input.txt'

def read_input (file_name) :
    file = open(file_name, 'r').read().split("\n")
    points = []
    for line in file :
        points.append([int(coord) for coord in line.split(',')])
    return points

def get_map_bounds (points) :
    max_x = max([point[0] for point in points]) + 3
    max_y = max([point[1] for point in points]) + 2
    return max_x, max_y

def get_map (points) :
    maxX, maxY = get_map_bounds(points)
    map = []
    for y in range(maxY + 2) :
        map.append(['.' for x in range(maxX)])
    return map

def get_map_with_points (points) :
    grid = get_map(points)
    for point in points :
        grid[point[1]][point[0]] = '#'
    return grid

def print_map (map) :
    for row in map :
        print(''.join(row))

def compute_area (point1, point2) :
    # Area include point lines
    return (abs(point1[0] - point2[0]) + 1) * (abs(point1[1] - point2[1]) + 1)

def print_area_on_map (map, point1, point2) :
    for x in range(abs(point1[0] - point2[0]) + 1) :
        for y in range(abs(point1[1] - point2[1]) + 1) :
            map[min(point1[1], point2[1]) + y][min(point1[0], point2[0]) + x] = 'O'
    print_map(map)

def test () :
    for i in range(len(points)) :
        segment = (points[i], points[i-1])
        if segment[0][0] == segment[1][0] :
            print("Vertical segment :", segment)
        elif segment[0][1] == segment[1][1] :
            print("Horizontal segment :", segment)
        else :
            print("Error")
            exit(1)

points = read_input(file_name)
maxArea = 0
maxAreaPoints = None

print(len(points))
# for point in points :
#     for other_point in points :
#         area = compute_area(point, other_point)
#         if area > maxArea : 
#             maxArea = area
#             maxAreaPoints = (point, other_point)

print("Max area :", maxArea, "between points", maxAreaPoints)
```

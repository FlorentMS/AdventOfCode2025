```python
fileName = 'input.txt'
primaryNumbers = [2, 3, 5, 7, 11, 13, 17, 19]
invalidIdSum = 0

def read_input (fileName) :
    file = open(fileName, 'r').read().split(",")
    ids = []
    for line in file :
        ids.append([int(id)for id in line.split('-')])
    return ids

def split_number_in(stringNumber, partSize, partNumber) :
    numberParts = []
    # print(f'Split "{stringNumber}" in {partNumber} by part of {partSize}')
    # Use a +1 to add last part of the splitted string
    for i in range(partNumber) :
        numberParts.append(stringNumber[:partSize])
        stringNumber = stringNumber[partSize:]
        # print(f'NumberParts = {numberParts}')
        # print(f'New string {stringNumber}')
    return numberParts

def check_equality_between_parts (parts) :
    for i in range(1, len(parts)) :
        if parts[i-1] != parts[i] :
            return False
    return True

def search_invalid_ids (ids) :
    global invalidIdSum
    for id in range(ids[0], ids[1] + 1) :
        idStr = str(id)
        idLength = len(idStr)
        for primaryNumber in primaryNumbers :
            if primaryNumber <= idLength :
                if idLength % primaryNumber == 0 :
                    # print(f'Primary number ({primaryNumber}) can divide lenght of {id}')
                    numberParts = split_number_in(idStr, idLength//primaryNumber, primaryNumber)
                    isInvalidId = check_equality_between_parts(numberParts)
                    if isInvalidId :
                        print(f'Id {id} is invalid')
                        invalidIdSum += id
                        break
            else :
                break

idsScopes = read_input(fileName)
print(idsScopes)
for scopes in idsScopes :
    search_invalid_ids(scopes)


print(f'Invalid ids sum is {invalidIdSum}')
```

```python
file_name = 'input.txt'
numbers = []
operations = []
results = []


def read_input (file_name) :
    file = open(file_name, 'r').read().split("\n")
    map = []
    mapOperation = [[],[],[]]
    mapOperator = []
    indexLastNumber = 0
    fileLenght = len(file)
    fileLenghtWithoutOperators = fileLenght - 1

    for line in file :
        map.append(list(line))

    mapLenght = len(map[0])
    lastIndex = mapLenght - 1

    i = 0
    while indexLastNumber < mapLenght :
        startIndexNumber = indexLastNumber
        endIndexNumber = 0
        if (i >= lastIndex) :
            endIndexNumber = lastIndex + 1
            for i in range(fileLenghtWithoutOperators - 1) :
                mapOperation[i].append(map[i][startIndexNumber:endIndexNumber])
            mapOperator.append(map[fileLenghtWithoutOperators][startIndexNumber])
            break
        if (map[0][i] == ' ' and map[1][i] == ' ' and map[2][i] == ' ' and map[3][i] == ' ') :
            endIndexNumber = i
            for i in range(fileLenghtWithoutOperators - 1) :
                mapOperation[i].append(map[i][startIndexNumber:endIndexNumber])
            mapOperator.append(map[fileLenghtWithoutOperators][startIndexNumber])
            indexLastNumber = i + 1
        i += 1
    return mapOperation, mapOperator


def get_numbers_to_operate (line1, line2, line3) :
    numbers = []
    for e in range(len(line1)) :
        number = ''
        if line1[e] != ' ' :
            number += line1[e]
        if line2[e] != ' ' :
            number += line2[e]
        if line3[e] != ' ' :
            number += line3[e]
        numbers.append(int(number))
    return numbers


print('Read input file:', file_name)
numbers, operations = read_input(file_name)
print('Numbers:', numbers)
print('Operations:', operations)

for map in numbers :
    if len(map) != len(operations) :
        print('Error: Number of operations does not match number of values')
        exit(1)

operationPointer = 0
for operation in operations :
    numbersToCompute = get_numbers_to_operate(numbers[0][operationPointer], numbers[1][operationPointer], numbers[2][operationPointer])
    if operation == '*' :
        result = 1
        for number in numbersToCompute :
            result *= number
    elif operation == '+' : 
        result = sum(numbersToCompute)
    results.append(result)
    operationPointer += 1

print('Results:', results)
print('Final Result:', sum(results))
```

```python
file_name = 'input.txt'
rollOfPaper = '@'

def read_input (file_name, detailsLog = False) :
    print('Read input file:', file_name)
    file = open(file_name, 'r').read().split("\n")
    formatedFile = []
    for line in file :
        formatedFile.append(list(line))
        if detailsLog :
            print('Line:', line)
    return formatedFile

def is_in_map(x, y):
    return 0 <= x < max_x and 0 <= y < max_y

def map_getter_adjacents(x,y) :
    adjacent = [
        "corner_top_left",
        "center_top",
        "corner_top_right",
        "center_right",
        "corner_bottom_right",
        "center_bottom",
        "corner_bottom_left",
        "center_left"
    ]
    position = {
        adjacent[0] : (x-1,y-1),
        adjacent[1] : (x,y-1),
        adjacent[2] : (x+1,y-1),
        adjacent[3] : (x+1,y),
        adjacent[4] : (x+1,y+1),
        adjacent[5] : (x,y+1),
        adjacent[6] : (x-1,y+1),
        adjacent[7] : (x-1,y)
    }
    return adjacent, position

def lookup_adjacents_positions(x, y):
    adjacentsCount = 0
    adjacentNames, adjacentCoordinates = map_getter_adjacents(x,y)
    # print('Looking up adjacents for x:', x, 'y:', y)
    for adjacentName in adjacentNames:
        # print('Checking adjacent:', adjacentName, 'at position', adjacentCoordinates[adjacentName])
        if is_in_map(adjacentCoordinates[adjacentName][0], adjacentCoordinates[adjacentName][1]) :
            if map[adjacentCoordinates[adjacentName][1]][adjacentCoordinates[adjacentName][0]] == rollOfPaper:
                adjacentsCount += 1
    return adjacentsCount


map = read_input(file_name)
max_y = len(map)
max_x = len(map[0])
accessibleRollsOfPapersCount = 0
for y in range(len(map)):
    for x in range(len(map[y])):
        if map[y][x] == rollOfPaper:
            adjacentCount = lookup_adjacents_positions(x,y)
            if adjacentCount < 4 :
                # print('Point x:', x, 'y:', y, 'is an accesible roll of paper', 'with', adjacentCount, 'adjacents')
                accessibleRollsOfPapersCount += 1
                

print('Total accessible rolls of paper:', accessibleRollsOfPapersCount)
```

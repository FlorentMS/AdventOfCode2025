```python
file_name = 'input.txt'

lightIndicatorsOn = '#'
lightIndicatorsOff =  '.'
buttonsWiringCountTotal = 0

def read_input (file_name, detailsLog = False) :
    print('Read input file:', file_name)
    file = open(file_name, 'r').read().split("\n")
    formatedLines = []
    if (detailsLog) :
        print('Elements :')
    for line in file :
        elements = line.split(' ')
        if (detailsLog) :
            print(elements)
        formatedLines.append([[],[],[]])

        ## Format line data ##
        for i in range(len(elements)) :
            cleanedElem = elements[i][1:-1] # Remove first and last character
            #print('cleanedElem', i, ':', cleanedElem)
            if (i == 0) :
                # light indicators
                formatedLines[-1][0] = bool_format_light(cleanedElem)
            elif (i == len(elements) - 1) :
                # joltage requirements
                formatedLines[-1][2] = [int(x) for x in cleanedElem.split(',')]
            else :
                # buttons wiring
                formatedLines[-1][1].append([int(x) for x in cleanedElem.split(',')])

    if (detailsLog) :
        print('Formated lines :')
        for line in formatedLines :
            print(line)
    return formatedLines

def use_button(boolLightMap, buttons) :
    resultMap = boolLightMap.copy()
    for index in buttons :
        resultMap[index] = not(boolLightMap[index])
    return resultMap

def pretty_format_light(boolLightMap, withBorder = False) :
    prettyMap = ''
    for bool in boolLightMap :
        if (bool) :
            prettyMap += lightIndicatorsOn
        else :
            prettyMap += lightIndicatorsOff
    if (withBorder) :
        return '[' + prettyMap + ']'
    else :
        return prettyMap
    
def bool_format_light(prettyLightMap, withBorder = False) :
    if (withBorder) :
        prettyLightMap = prettyLightMap[1:-1]
    return [x == lightIndicatorsOn for x in prettyLightMap]

def resolve_puzzle() :
    buttonsWiringCount = 1
    findLightCombination = False
    lightIndicators = [False for x in lightGoal]
    if (lightIndicators == lightGoal) :
        return 0, []
    while buttonsWiringCount < 100 and not(findLightCombination) :
        result = test_for_buttons(lightIndicators, 1, buttonsWiringCount)
        # print('Resultat for', buttonsWiringCount, 'buttons :', result)
        if (result is not None) :
            findLightCombination = True
            return buttonsWiringCount, result
        buttonsWiringCount += 1
    return None, None

def test_for_buttons(lightIndicators, deepness, deepnessMax) :
    # print('Testing deepness', deepness, 'of max', deepnessMax)
    for button in buttons :
        result = use_button(lightIndicators, button)
        # print('Current light state :', pretty_format_light(lightIndicators, True))
        # print('Result', pretty_format_light(result, True), 'after using button', button)
        if (lightGoal == result) :
            print('Goal reached !')
            print(pretty_format_light(result, True))
            return [button]
        if (deepness < deepnessMax) :
            resultDeepest = test_for_buttons(result, deepness + 1, deepnessMax)
            if (resultDeepest is not None) :
                return [button] + resultDeepest
    return None

def test_solution(buttonsCombination, lightGoal) :
    lightIndicators = [False for x in lightGoal]
    for button in buttonsCombination :
        lightIndicators = use_button(lightIndicators, button)
    print('Comparison (result, goal):', lightIndicators, lightGoal)
    return lightIndicators == lightGoal

    

machinesManualLines = read_input(file_name)

for instruction in machinesManualLines :
    lightGoal = instruction[0]
    print('Goal', pretty_format_light(lightGoal, True))
    buttons = instruction[1]

    buttonsCount, buttonsCombination = resolve_puzzle()
    if (buttonsCombination is not None) :
        if (test_solution(buttonsCombination, lightGoal)) :
            print('Solution found with', buttonsCount, 'buttons pressed :')
            for button in buttonsCombination :
                print(button)
            buttonsWiringCountTotal += buttonsCount
    else :
        print('No solution found')

print('Total buttons pressed for this machine :', buttonsWiringCountTotal)
```

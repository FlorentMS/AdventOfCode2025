```python
file_name = 'input.txt'

lightIndicatorsOn = '#'
lightIndicatorsOff =  '.'
buttonsWiringCountTotal = 0

def read_input (file_name, detailsLog = False) :
    print('Read input file:', file_name)
    file = open(file_name, 'r').read().split("\n")
    formatedLines = []
    if (detailsLog) :
        print('Elements :')
    for line in file :
        elements = line.split(' ')
        if (detailsLog) :
            print(elements)
        formatedLines.append([[],[],[]])

        ## Format line data ##
        for i in range(len(elements)) :
            cleanedElem = elements[i][1:-1] # Remove first and last character
            #print('cleanedElem', i, ':', cleanedElem)
            if (i == 0) :
                # light indicators
                formatedLines[-1][0] = bool_format_light(cleanedElem)
            elif (i == len(elements) - 1) :
                # joltage requirements
                formatedLines[-1][2] = [int(x) for x in cleanedElem.split(',')]
            else :
                # buttons wiring
                formatedLines[-1][1].append([int(x) for x in cleanedElem.split(',')])

    if (detailsLog) :
        print('Formated lines :')
        for line in formatedLines :
            print(line)
    return formatedLines

def use_button(joltagesMap, buttons) :
    resultMap = joltagesMap.copy()
    for index in buttons :
        resultMap[index] += 1
    return resultMap

def pretty_format_light(boolLightMap, withBorder = False) :
    prettyMap = ''
    for bool in boolLightMap :
        if (bool) :
            prettyMap += lightIndicatorsOn
        else :
            prettyMap += lightIndicatorsOff
    if (withBorder) :
        return '[' + prettyMap + ']'
    else :
        return prettyMap
    
def bool_format_light(prettyLightMap, withBorder = False) :
    if (withBorder) :
        prettyLightMap = prettyLightMap[1:-1]
    return [x == lightIndicatorsOn for x in prettyLightMap]

def resolve_puzzle_old() :
    buttonsWiringCount = min(joltageRequirement)
    findJoltageCombination = False
    joltages = [0 for x in joltageRequirement]
    if (joltages == joltageRequirement) :
        return 0, []
    while buttonsWiringCount < sum(joltageRequirement) and not(findJoltageCombination) :
        result = test_for_buttons_old(joltages, 1, buttonsWiringCount)
        # print('Resultat for', buttonsWiringCount, 'buttons :', result)
        if (result is not None) :
            findJoltageCombination = True
            return buttonsWiringCount, result
        buttonsWiringCount += 1
    return None, None

def test_for_buttons_old(joltages, deepness, deepnessMax) :
    # print('Testing deepness', deepness, 'of max', deepnessMax)
    # print('.', end =" ")
    no_previous_joltage_overflow = not(is_joltage_overflow(joltages))
    for button in buttons :
        result = use_button(joltages, button)

        # print('Current light state :', pretty_format_light(lightIndicators, True))
        # print('Result', pretty_format_light(result, True), 'after using button', button)
        if (joltageRequirement == result) :
            print('Goal reached :', result)
            return [button]
        if (no_previous_joltage_overflow and deepness < deepnessMax) :
            resultDeepest = test_for_buttons(result, deepness + 1, deepnessMax)
            if (resultDeepest is not None) :
                return [button] + resultDeepest
    return None

def resolve_puzzle() :
    global joltagesResults
    buttonsWiringCount = 1
    findJoltageCombination = False
    joltagesResults = [start_joltages()]
    if (joltagesResults == joltageRequirement) :
        findJoltageCombination = True
        return 0, []
    while buttonsWiringCount < sum(joltageRequirement) and not(findJoltageCombination) :
        #print ('joltagesResults', joltagesResults)
        joltages = joltagesResults.copy()
        joltagesResults.clear()
        for joltage in joltages :
            result = test_for_buttons(joltage, buttonsWiringCount)
            # print('Resultat for', buttonsWiringCount, 'buttons :', result)
            if (result is not None) :
                findJoltageCombination = True
                return buttonsWiringCount
        buttonsWiringCount += 1
    return None

def test_for_buttons(joltages, deepness) :
    global joltagesResults
    for button in buttons :
        result = use_button(joltages, button)
        if (not(is_joltage_overflow(result))) :
            joltagesResults.append(result)

        if (joltageRequirement == result) :
            print('Goal reached :', result)
            return deepness
    return None


def start_joltages() :
    return [0 for x in joltageRequirement]

# def test_solution(buttonsCombination, joltageRequirement) :
#     joltages = start_joltages()
#     for button in buttonsCombination :
#         joltages = use_button(joltages, button)
#     print('Comparison (result, goal):', joltages, joltageRequirement)
#     return joltages == joltageRequirement

def is_joltage_overflow(joltages) :
    index = 0
    for joltage in joltages :
        if joltage > joltageRequirement[index] :       
            # print('OverflowError')
            return True
        index += 1
    return False
    

machinesManualLines = read_input(file_name)

for instruction in machinesManualLines :
    lightGoal = instruction[0]
    joltageRequirement = instruction[2]
    print('Goal', joltageRequirement)
    buttons = instruction[1]

    results = []
    buttonsCount = resolve_puzzle()
    if (buttonsCount is not None) :
        print('Solution found with', buttonsCount, 'buttons pressed :')
        buttonsWiringCountTotal += buttonsCount
    else :
        print('No solution found')
        exit(1)
    # break

print('Total buttons pressed for this machine :', buttonsWiringCountTotal)
```

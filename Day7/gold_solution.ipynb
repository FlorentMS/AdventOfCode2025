```python
file_name = 'input.txt'
tachyonBeamSymbol = '|'
splitterSymbol = '^'
startSymbol = 'S'

def read_input (file_name) :
    file = open(file_name, 'r').read().split("\n")
    return [list(line) for line in file]

# Use 0 instead of '.' for easier calculations of timelines
def transform_map (map) :
    for y in range(len(map)) :
        for x in range(len(map[y])) :
            if map[y][x] == '.' :
                map[y][x] = int(0)
    return map

def tachyon_propagation () :
    global map
    for y in range(1, len(map) - 1) :
        # print("Processing row:", y)
        for x in range(0, len(map[y])) :
            # print(" Processing column:", x, "Symbol:", map[y][x])
            nextRow = map[y + 1]
            if map[y][x] != splitterSymbol :
                if map[y][x] > 0 :
                    if nextRow[x] == splitterSymbol:
                        nextRow[x-1] += map[y][x]
                        nextRow[x+1] += map[y][x]
                    else :
                        nextRow[x] += map[y][x]

map = read_input(file_name)
indexStartPoint = len(map[0])//2
if map[0][indexStartPoint] == startSymbol :
    map[1][indexStartPoint] = 1
else :
    print("Error: No starting point found.")
    exit(1)

map = transform_map(map)
tachyon_propagation()

lastLineInMap = map[-1]
timelinePossibilities = 0
for elem in lastLineInMap :
    timelinePossibilities += elem
print("Tachyon Beam possibles timelines:", timelinePossibilities)
```

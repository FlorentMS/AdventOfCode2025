```python
file_name = 'input.txt'
batteries = []
turnedOnBatteries = []
relative_next_index = []
voltage = 0
maxTurnedOnBatteries = 12

def find_highest (list, max = 9) :
    maxList = 0
    index_maxList = 0
    for i in range(len(list)) :
        if (list[i] > maxList) :
            if (list[i] <= max) :
                maxList = list[i]
                index_maxList = i
    return maxList, index_maxList

def find_two_highest (bank) :
    sec_highest = 0
    i = 9
    while (sec_highest == 0) and (i > 0) :
        first_highest, first_index = find_highest(bank, i)
        # print('first (', first_highest, first_index, ')')
        sec_highest, sec_index = find_highest(bank[first_index+1:])
        sec_index += first_index+1
        # print('second (', sec_highest, sec_index, ')')
        i -= 1
    return first_highest, first_index, sec_highest, sec_index

def search_for_next_highest (bank, n_highest) :
    global relative_next_index
    next_highest = 0
    i = 9
    # print('Search next in bank', bank)
    # print('Need to find', n_highest, 'more highest batteries')
    while (next_highest == 0) and (i > 0) and (n_highest >= 0) :
        highest, index = find_highest(bank, i)
        # print('Found highest', highest, 'at index', index)
        subBank = bank[index+1:]
        if  len(subBank) >=  n_highest - 1 and n_highest > 1:
            turnedOnBatteries.append(highest)
            relative_next_index.append(index)
            # print('Found candidate', highest, 'at index', index, 'create subBank', subBank) 
            search_for_next_highest(subBank, n_highest - 1)
            return 1
        elif n_highest == 1:
            turnedOnBatteries.append(highest)
            relative_next_index.append(index)
            # print('Found last candidate', highest, 'at index', index) 
            return 1
        i -= 1
    return 1

def find_highests (bank) :
    print('Find highests in bank', bank)
    search_for_next_highest (bank, maxTurnedOnBatteries)

file = open(file_name, 'r').read().split("\n")
for line in file:
    batteries.append([int(i) for i in list(line)])


print("Start")

for bank in batteries:
    turnedOnBatteries = []
    relative_next_index = []
    # print('bank', bank)
    # print(find_two_highest (bank))

    find_highests (bank)
    print('Turned on batteries', turnedOnBatteries)
    print('Relative indexes', relative_next_index)
    bankVoltage = 0
    bankVoltageStr = ''
    for battery in turnedOnBatteries:
        bankVoltageStr += str(battery)
    print('Bank voltage', bankVoltageStr)
    voltage += int(bankVoltageStr)

print("Final voltage", voltage)
```
